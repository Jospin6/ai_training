```python
def two_sum(nums: list[int], target: int) -> list[int]:

    seen = {}

    for i, v in enumerate(nums):
        complement = target - v

        if complement in seen:
            return [seen[complement], i]

        seen[v] = i
```

```python
nums = [2,7,11,15] 
target = 9

print(two_sum(nums, target))
```

```python
def BTBSS(prices: list[int]) -> int:
    min_price = float("inf")
    max_profit = float("-inf")

    for price in prices:
        min_price = min(min_price, price)
            
        today_profit = price - min_price

        max_profit = max(max_profit, today_profit)

    return max_profit
```

```python
prices = [7,1,5,3,6,4]
print(BTBSS(prices))
```

```python
def CD(nums: list[int]) -> bool:
    # can we fix this using a set() ?

    seen = {}

    for i, v in enumerate(nums):
        if v in seen:
            return True
        seen[v] = i

    return False
```

```python
nums = [1,2,3,1]
print(CD(nums))
```

```python
def productExceptSelf(nums: list[int]) -> list[int]:
        n = len(nums)
        answer = [1] * n

        # Produits des préfixes (gauche)
        prefix = 1
        for i in range(n):
            answer[i] = prefix
            prefix *= nums[i]

        # Produits des suffixes (droite)
        suffix = 1
        for i in range(n - 1, -1, -1):
            answer[i] *= suffix
            suffix *= nums[i]

        return answer
```

```python
print(productExceptSelf([1,2,3,4]))
```

```python
def maxSubArray(nums: list[int]) -> int:
        current_sum = nums[0]
        max_sum = nums[0]

        for i in range(1, len(nums)):
            current_sum = max(nums[i], current_sum + nums[i])
            max_sum = max(max_sum, current_sum)

        return max_sum
```

```python
def maxProduct(nums: list[int]) -> int:
        max_prod = nums[0]
        min_prod = nums[0]
        result = nums[0]

        for i in range(1, len(nums)):
            if nums[i] < 0:
                max_prod, min_prod = min_prod, max_prod

            max_prod = max(nums[i], max_prod * nums[i])
            min_prod = min(nums[i], min_prod * nums[i])

            result = max(result, max_prod)

        return result
```

```python
nums = [1,4,6,-4,8,3]
print(maxProduct(nums))
```

```python
def findMin(nums: list[int]) -> int:
        left = 0
        right = len(nums) -1
        
        while left < right:
            mid = (left + right)//2

            if nums[mid] > nums[right]:
                left = mid + 1
            else:
                right = mid

        return nums[left]
```

```python
def search(nums: list[int], target: int) -> int:
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid

            # La partie gauche est triée
            if nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1

            # Sinon, la partie droite est triée
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1

        return -1
```

```python
def threeSum(nums: list[int]) -> list[list[int]]:
        nums.sort()
        result = []

        for i in range(len(nums)):

            # Ignorer les doublons
            if i > 0 and nums[i] == nums[i - 1]:
                continue

            left = i + 1
            right = len(nums) - 1

            while left < right:
                total = nums[i] + nums[left] + nums[right]

                if total > 0:
                    right -= 1

                elif total < 0:
                    left += 1

                else:
                    result.append([nums[i], nums[left], nums[right]])

                    left += 1

                    while left < right and nums[left] == nums[left - 1]:
                        left += 1

        return result
```

```python
def maxArea(height: list[int]) -> int:
        left = 0
        right = len(height) - 1

        max_area = 0

        while left < right:

            area = (right - left) * min(height[left], height[right])

            max_area = max(max_area, area)

            if height[left] < height[right]:
                left += 1
            else:
                right -= 1

        return max_area
```
